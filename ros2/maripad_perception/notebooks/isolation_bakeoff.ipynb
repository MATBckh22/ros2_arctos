# MARIPAD Isolation Method Bakeoff

Capture 20 RGB-D frames first, then use this notebook to compare the depth, HSV, and background-subtraction masks before choosing the production isolation method.

```python
from pathlib import Path
import cv2
import matplotlib.pyplot as plt
from maripad_perception.perception.book_detector import BookDetector

DATASET_DIR = Path.home() / 'maripad_data' / 'book_dataset'
detector = BookDetector()
frame_dirs = sorted(path for path in DATASET_DIR.iterdir() if path.is_dir())[:20]
len(frame_dirs)
```

```python
def load_frame(frame_dir):
    rgb_bgr = cv2.imread(str(frame_dir / 'rgb.png'), cv2.IMREAD_COLOR)
    depth = cv2.imread(str(frame_dir / 'depth.png'), cv2.IMREAD_UNCHANGED)
    return cv2.cvtColor(rgb_bgr, cv2.COLOR_BGR2RGB), depth

frame_dir = frame_dirs[0]
rgb, depth = load_frame(frame_dir)
masks = detector.generate_masks_for_all_methods(rgb, depth)
metrics = {name: detector.score_mask_quality(mask) for name, mask in masks.items()}
metrics
```

```python
fig, axes = plt.subplots(1, 4, figsize=(18, 5))
axes[0].imshow(rgb)
axes[0].set_title('RGB')
axes[0].axis('off')
for axis, method in zip(axes[1:], ['depth', 'hsv', 'background']):
    axis.imshow(masks[method], cmap='gray')
    axis.set_title(f"{method}\n{metrics[method]}")
    axis.axis('off')
plt.tight_layout()
```
